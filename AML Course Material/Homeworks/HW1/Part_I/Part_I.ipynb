```python
point = torch.tensor([5.0, 6.0, 8.0, 9.0], requires_grad = True)
coeffs = torch.tensor([1.0, 3.0, 5.0, 6.0])

def g(p):
    return torch.sum(coeffs * torch.exp(p) * p**2)

function_values = g(point)
function_values.backward()
gradient = point.grad
print(gradient)
```

```python
A = torch.tensor([[4.0, 3.0], [7.0, 9.0]], requires_grad = True)

B = torch.tensor([[3.0, 5.0], [1.0, 11.0]])

inside = A.T @ A @ B.T @ A @ A.T @ A @ B
f = torch.log(torch.norm(inside, p=2)**2)

print(f)
```
